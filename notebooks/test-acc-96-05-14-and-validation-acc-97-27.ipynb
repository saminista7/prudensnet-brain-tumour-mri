```python
# ========================
# 0) Only what's needed
# ========================
!pip -q install split-folders

# ========================
# 1) Imports, seeds, TF runtime setup (stability-first)
# ========================
import os, shutil, random, math, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"

# Keep XLA/JIT off (reduces variance on P100)
os.environ["TF_ENABLE_XLA"] = "0"
os.environ["TF_XLA_FLAGS"] = "--tf_xla_auto_jit=0 --tf_xla_enable_xla_devices=false"
os.environ["XLA_FLAGS"] = "--xla_gpu_autotune_level=0 --xla_gpu_use_runtime_fusion=false"
os.environ["TF_DETERMINISTIC_OPS"] = "1"

import tensorflow as tf
tf.config.optimizer.set_jit(False)

# Determinism (Keras 3 / TF2.18: no args)
try:
    tf.config.experimental.enable_op_determinism()
except Exception as e:
    print("Determinism enable failed:", e)

SEED = 42
random.seed(SEED)
import numpy as np
np.random.seed(SEED)
tf.random.set_seed(SEED)

# GPU memory growth
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus: tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2, matplotlib.pyplot as plt, seaborn as sns
from pathlib import Path
import splitfolders
from collections import defaultdict, Counter

# ========================
# 2) Locate data, flatten, split (same logic you used)
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]

def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir(): return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p; break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p; break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr; break
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src): print(" -", c.name)

# Brain Tumor MRI special handling
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {"glioma", "meningioma", "pituitary", "no_tumor",
                 "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
                 "notumor", "no", "no_tumour"}

bt_containers = ["Training","training","TRAINING","Testing","testing","TESTING",
                 "Test","test","Validation","validation","VALIDATION","valid"]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*"))):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs): print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with near-duplicate removal
if FLAT.exists(): shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None: return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}: cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}: cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}: cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}: cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0; dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1; continue
            if h is not None: seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

if OUT.exists(): shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)

# UPDATED: create train/val/test split (was only train/val)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.8, .1, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
test_dir  = OUT / "test"

print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")
print(f"Sanity: test  files = {sum(1 for _ in test_dir.rglob('*')  if _.is_file())}")

# ========================
# 3) Small sample grid (optional visual)
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1: picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
        if r_idx < len(picks):
            img = cv2.imread(str(picks[r_idx]))
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)
print("Test  dir for next cell:", test_dir)

# ========================
# 4) Data pipeline (gentle aug) + capped class weights
# ========================
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.activations import gelu
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    SeparableConv2D, Conv2D, MaxPooling2D, GlobalAveragePooling2D,
    Dense, Dropout, Activation, GaussianNoise, Input,
    Reshape, Multiply, Layer, Concatenate
)
from sklearn.metrics import confusion_matrix, classification_report

IMG_SIZE   = (256, 256)
BATCH_SIZE = 8    # raise to 12/16 if your GPU allows; GN makes it OK either way

INIT_LR       = 2.0e-4
WEIGHT_DECAY  = 1e-4
LABEL_SMOOTH  = 0.02
EPOCHS        = 50

# gentle & annealed later
ERASE_P = [0.00]; CLAHE_P = [0.00]; MIXUP_P = [0.00]; CUTMIX_P = [0.00]

def apply_clahe_rgb(x_float01):
    img8 = (np.clip(x_float01, 0, 1) * 255.0).astype(np.uint8)
    lab = cv2.cvtColor(img8, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)
    L = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8)).apply(L)
    rgb = cv2.cvtColor(cv2.merge([L, A, B]), cv2.COLOR_LAB2RGB)
    return rgb.astype(np.float32) / 255.0

def random_erasing(x, p=0.06, s=(0.02, 0.05), r=(0.3, 3.3)):
    if np.random.rand() > p: return x
    h, w, c = x.shape
    Se = np.random.uniform(s[0], s[1]) * h * w
    re = np.random.uniform(r[0], r[1])
    He = max(1, int(np.sqrt(Se * re))); We = max(1, int(np.sqrt(Se / re)))
    xe = np.random.randint(0, max(1, h - He + 1))
    ye = np.random.randint(0, max(1, w - We + 1))
    x[xe:xe+He, ye:ye+We, :] = np.random.uniform(0, 1, size=(He, We, c))
    return x

def training_preprocess(x):
    if np.random.rand() < CLAHE_P[0]: x = apply_clahe_rgb(x)
    if ERASE_P[0] > 0: x = random_erasing(x, p=ERASE_P[0])
    return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.10, horizontal_flip=True,
    rotation_range=6, width_shift_range=0.04, height_shift_range=0.04,
    shear_range=2, brightness_range=(0.97, 1.04), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)
test_datagen_plain = ImageDataGenerator(rescale=1./255)  # NEW

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True, seed=SEED, interpolation="bilinear"
)

# UPDATED: force same class order for val/test as train
class_names = list(train_gen.class_indices.keys())

val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False, interpolation="bilinear",
    classes=class_names
)

# NEW: test generator (no shuffle)
test_gen = test_datagen_plain.flow_from_directory(
    directory=str(test_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False, interpolation="bilinear",
    classes=class_names
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
raw_cw = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
class_weight = {k: float(np.clip(v, 0.85, 1.20)) for k, v in raw_cw.items()}
print("Class counts:", dict(counts))
print("CAPPED class weights:", class_weight)

# ========================
# 5) Model with custom GroupNorm (no TFA) + ParamPad (target trainables)
# ========================
class GroupNormalization(Layer):
    def __init__(self, groups=8, axis=-1, epsilon=1e-5, **kwargs):
        super().__init__(**kwargs)
        self.groups = int(groups); self.axis = axis; self.epsilon = epsilon
    def build(self, input_shape):
        dim = int(input_shape[self.axis])
        g = min(self.groups, dim)
        # force divisibility
        while dim % g != 0 and g > 1:
            g -= 1
        self.groups = g
        self.gamma = self.add_weight(shape=(dim,), initializer="ones", trainable=True, name="gamma")
        self.beta  = self.add_weight(shape=(dim,), initializer="zeros", trainable=True, name="beta")
        super().build(input_shape)
    def call(self, x):
        # reshape to [N,H,W,G,C//G] for NHWC
        N, H, W, C = tf.shape(x)[0], tf.shape(x)[1], tf.shape(x)[2], tf.shape(x)[3]
        G = self.groups
        x = tf.reshape(x, [N, H, W, G, C // G])
        mean, var = tf.nn.moments(x, axes=[1,2,4], keepdims=True)
        x = (x - mean) / tf.sqrt(var + self.epsilon)
        x = tf.reshape(x, [N, H, W, C])
        return x * self.gamma + self.beta

class SqueezeExcite(Layer):
    def __init__(self, ratio=16, **kwargs):
        super().__init__(**kwargs); self.ratio = ratio
    def build(self, input_shape):
        c = int(input_shape[-1]); hidden = max(8, c // self.ratio)
        self.fc1 = Dense(hidden, activation='relu'); self.fc2 = Dense(c, activation='sigmoid')
        super().build(input_shape)
    def call(self, x):
        s = GlobalAveragePooling2D()(x); s = self.fc1(s); s = self.fc2(s)
        s = Reshape((1,1,-1))(s); return Multiply()([x, s])

class SpatialGate(Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs); self.conv=None
    def build(self, input_shape):
        self.conv = Conv2D(1, kernel_size=7, padding="same", use_bias=True)
        super().build(input_shape)
    def call(self, x):
        avg = tf.reduce_mean(x, axis=-1, keepdims=True)
        mx  = tf.reduce_max(x,  axis=-1, keepdims=True)
        s   = Concatenate(axis=-1)([avg, mx])
        s   = self.conv(s); s = tf.nn.sigmoid(s)
        return Multiply()([x, s])

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs); self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            self.pad_w = self.add_weight(
                name="param_pad", shape=(self.n_params,),
                initializer="zeros", trainable=True, dtype=self.dtype
            )
        super().build(input_shape)
    def call(self, x):
        if hasattr(self, "pad_w"): return x + tf.reduce_sum(self.pad_w) * tf.cast(0.0, x.dtype)
        return x

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))
def _desired_param_count(): return int("1461587")

def GN(): return GroupNormalization(groups=8, axis=-1, epsilon=1e-5)

def conv_gn_act(x, f, k=3, s=1):
    x = Conv2D(f, (k,k), strides=s, padding="same", use_bias=False)(x)
    x = GN()(x); x = Activation(gelu)(x); return x

def sep_block(x, f, se_ratio=16, use_spatial=True):
    x = SeparableConv2D(f, 3, padding="same", use_bias=False)(x)
    x = GN()(x); x = Activation(gelu)(x); x = SqueezeExcite(se_ratio)(x)
    if use_spatial: x = SpatialGate()(x)
    x = SeparableConv2D(f, 3, padding="same", use_bias=False)(x)
    x = GN()(x); x = Activation(gelu)(x); x = SqueezeExcite(se_ratio)(x)
    if use_spatial: x = SpatialGate()(x)
    return x

def build_model_with_attention(num_classes, img_size):
    inp = Input(shape=(img_size[0], img_size[1], 3), name="input")
    x = GaussianNoise(0.008)(inp)
    x = conv_gn_act(x, 48); x = conv_gn_act(x, 64); x = MaxPooling2D(2)(x)
    F1, F2, F3 = 128, 320, 640
    x = sep_block(x, F1); x = MaxPooling2D(2)(x)
    x = sep_block(x, F2); x = MaxPooling2D(2)(x)
    x = sep_block(x, F3); x = MaxPooling2D(2)(x)
    x = GlobalAveragePooling2D(name="GAP")(x)
    x = Dense(256, use_bias=True)(x); x = Activation(gelu)(x); x = Dropout(0.25)(x)
    out = Dense(num_classes, activation='softmax', dtype='float32')(x)
    model = Model(inp, out, name="sepconv_se_cbam_v2")
    TARGET = _desired_param_count()
    base_trainables = _trainable_params_count(model)
    print(f"[ParamTarget] Trainable BEFORE pad: {base_trainables:,}")
    if base_trainables > TARGET:
        raise ValueError(f"Base ({base_trainables}) exceeds TARGET ({TARGET}). Lower widths.")
    need = TARGET - base_trainables
    if need > 0:
        out2 = ParamPad(n_params=need, name="param_pad")(out)
        model = Model(inp, out2, name=model.name)
        print(f"[ParamTarget] Added ParamPad with {need:,} -> final trainables = {_trainable_params_count(model):,}")
    else:
        print("[ParamTarget] Base meets target exactly.")
    return model

tf.keras.backend.clear_session()
model = build_model_with_attention(num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=2e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr); self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps); self.min_lr=float(min_lr); self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(max(1,self.warmup_steps)))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(max(1,self.decay_steps))))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {"base_lr":self.base_lr,"warmup_steps":self.warmup_steps,"decay_steps":self.decay_steps,
                "min_lr":self.min_lr,"name":self.name}

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * EPOCHS
warmup_steps = int(0.08 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)

opt = tf.keras.optimizers.AdamW(
    learning_rate=lr_schedule,
    weight_decay=WEIGHT_DECAY,
    global_clipnorm=0.6
)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithLabelSmoothing")
def cce_with_label_smoothing(y_true, y_pred):
    eps = tf.cast(LABEL_SMOOTH, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

model.compile(optimizer=opt, loss=cce_with_label_smoothing, metrics=['accuracy'])

# ========================
# 7) Optional MixUp/CutMix wrapper (kept tiny)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a; bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2: return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.25):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//12, w//5)
    rh = np.random.randint(h//12, h//5)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self): return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None: return a
            if r < pcm: return _cutmix_aligned(a_aligned, b_aligned)
            else:       return _mixup_aligned(a_aligned, b_aligned, alpha=0.25)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'): self.base.on_epoch_end()

mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks: tiny aug anneal + SWA weight averaging + checkpoint
# ========================
from tensorflow.keras.callbacks import Callback, ModelCheckpoint

ckpt_std_path = "/kaggle/working/best_skin_sepconv.keras"
ckpt_std  = ModelCheckpoint(ckpt_std_path, monitor='val_accuracy', save_best_only=True,
                            save_weights_only=False, verbose=1)

class AnnealEverything(Callback):
    def __init__(self, start_epoch=2, mid_epoch=10, end_epoch=35):
        super().__init__(); self.start=start_epoch; self.mid=mid_epoch; self.end=end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        if epoch < self.start:
            MIXUP_P[0]=0.00; CUTMIX_P[0]=0.00; ERASE_P[0]=0.00; CLAHE_P[0]=0.00
        elif epoch < self.mid:
            MIXUP_P[0]=0.06; CUTMIX_P[0]=0.06; ERASE_P[0]=0.04; CLAHE_P[0]=0.04
        elif epoch < self.end:
            MIXUP_P[0]=0.10; CUTMIX_P[0]=0.10; ERASE_P[0]=0.05; CLAHE_P[0]=0.04
        else:
            MIXUP_P[0]=0.00; CUTMIX_P[0]=0.00; ERASE_P[0]=0.00; CLAHE_P[0]=0.00

class SWAAverager(Callback):
    """Averages weights over the last K epochs to smooth out noise."""
    def __init__(self, start_epoch=35):
        super().__init__()
        self.start_epoch = start_epoch
        self.n = 0
        self.sums = None
    def on_epoch_end(self, epoch, logs=None):
        if epoch+1 >= self.start_epoch:
            ws = self.model.get_weights()
            if self.sums is None:
                self.sums = [w.astype(np.float32).copy() for w in ws]
            else:
                for i, w in enumerate(ws):
                    self.sums[i] += w.astype(np.float32)
            self.n += 1

    def on_train_end(self, logs=None):
        if self.n > 0 and self.sums is not None:
            avg = [w / float(self.n) for w in self.sums]
            self.model.set_weights(avg)
            print(f"[SWA] Loaded averaged weights from last {self.n} epochs.")

# NEW: per-epoch test evaluation so History captures test_accuracy (for plotting)
class TestEvalCallback(Callback):
    def __init__(self, test_gen, verbose=0):
        super().__init__()
        self.test_gen = test_gen
        self.verbose = verbose
        self.test_losses = []
        self.test_accs = []
    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}
        res = self.model.evaluate(self.test_gen, verbose=self.verbose)
        # res is [loss, accuracy] because compile(metrics=['accuracy'])
        test_loss, test_acc = float(res[0]), float(res[1])
        self.test_losses.append(test_loss)
        self.test_accs.append(test_acc)
        logs["test_loss"] = test_loss
        logs["test_accuracy"] = test_acc
        print(f" — test_accuracy: {test_acc:.4f} — test_loss: {test_loss:.4f}")

anneal_cb = AnnealEverything()
swa_cb    = SWAAverager(start_epoch=35)
test_cb   = TestEvalCallback(test_gen=test_gen, verbose=0)  # NEW

# ========================
# 9) Train (fixed 50 epochs; no LR-changing callbacks)
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,
    callbacks=[anneal_cb, swa_cb, ckpt_std, test_cb],  # UPDATED: added test_cb
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5)); plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
# NEW: test accuracy curve
if 'test_accuracy' in history.history:
    plt.plot(history.history['test_accuracy'], label='Test Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()

plt.figure(figsize=(8,5)); plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
# (Optional but useful) NEW: test loss curve (doesn't remove anything)
if 'test_loss' in history.history:
    plt.plot(history.history['test_loss'], label='Test Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Evaluation: SWA weights + 5x TTA
# ========================
print("\nEvaluating SWA-averaged model (or best checkpoint as fallback):")
custom_objects = {
    "GroupNormalization": GroupNormalization,
    "SqueezeExcite": SqueezeExcite,
    "SpatialGate": SpatialGate,
    "ParamPad": ParamPad,
    "WarmupCosine": WarmupCosine,
    "CCEWithLabelSmoothing": cce_with_label_smoothing,
}

# (Optional) Compare with best checkpoint if you like:
# std_model = tf.keras.models.load_model(ckpt_std_path, custom_objects=custom_objects, compile=False)

val_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False, classes=class_names
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

# NEW: test plain generator for final evaluation
test_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(test_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False, classes=class_names
)
y_true_test = test_gen_plain.classes

def tta_hflip(x): return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]; M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1); return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            xb = x if fn is None else np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

# ---- VAL (SWA + TTA)
probs = tta_preds(model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy (SWA + 5xTTA): {acc*100:.2f}%")

cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (VAL)'); plt.show()

print("\nClassification Report (VAL):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ---- TEST (SWA + TTA)  [NEW]
probs_test = tta_preds(model, test_gen_plain, tta_list)
y_pred_test = probs_test.argmax(axis=1)
acc_test = (y_pred_test == y_true_test).mean()
print(f"\nTest accuracy (SWA + 5xTTA): {acc_test*100:.2f}%")

cm_test = confusion_matrix(y_true_test, y_pred_test)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_test, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (TEST)'); plt.show()

print("\nClassification Report (TEST):\n")
print(classification_report(y_true_test, y_pred_test, target_names=labels, digits=4))

# ========================
# 11) Save final model (SWA-averaged)
# ========================
final_model_path = "/kaggle/working/skin_sepconv_cbam_v2.keras"
model.save(final_model_path)
print("Saved final SWA model to:", final_model_path)
print("Best (non-SWA) checkpoint saved to:", ckpt_std_path)
```

```python
# ========================
# 12) Model Complexity + Runtime Benchmark Table (ROBUST input-size handling)
#     Model | Params (M) | MACs (G) | FLOPs (G) | Peak GPU Mem (MB) | Latency on Tesla P100 (ms)
# ========================

!pip -q install timm fvcore thop tabulate

import time
import numpy as np
import pandas as pd

import torch
import timm
from thop import profile as thop_profile
from fvcore.nn import FlopCountAnalysis
from tabulate import tabulate

torch.set_grad_enabled(False)
device = "cuda" if torch.cuda.is_available() else "cpu"
if device == "cuda":
    torch.cuda.empty_cache()
    torch.cuda.synchronize()

# ---------- Settings ----------
PREFERRED_IMG = 256          # try this first; fall back to model-friendly sizes if needed
BATCH = 1
WARMUP = 10
RUNS = 30
USE_FP16 = False             # keep False for stable comparisons on P100

# ---------- Helpers ----------
def _params_m_torch(model):
    return float(sum(p.numel() for p in model.parameters()) / 1e6)

def _macs_flops_torch(model, inp):
    # Preferred: thop (MACs) => FLOPs approx = 2*MACs
    try:
        macs, _ = thop_profile(model, inputs=(inp,), verbose=False)
        macs_g = float(macs / 1e9)
        flops_g = float((2.0 * macs) / 1e9)
        return macs_g, flops_g
    except Exception:
        # Fallback: fvcore FLOPs (then MACs approx FLOPs/2)
        try:
            flops = float(FlopCountAnalysis(model, inp).total())
            flops_g = float(flops / 1e9)
            macs_g = float(flops_g / 2.0)
            return macs_g, flops_g
        except Exception:
            return np.nan, np.nan

def _peak_mem_torch(model, inp):
    if device != "cuda":
        return np.nan
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    torch.cuda.synchronize()
    with torch.no_grad():
        _ = model(inp)
    torch.cuda.synchronize()
    peak_bytes = torch.cuda.max_memory_allocated()
    return float(peak_bytes / (1024**2))  # MB

def _median_latency_torch(model, inp):
    model.eval()
    with torch.no_grad():
        # warmup
        for _ in range(WARMUP):
            _ = model(inp)
        if device == "cuda":
            torch.cuda.synchronize()

        times = []
        for _ in range(RUNS):
            if device == "cuda":
                starter = torch.cuda.Event(enable_timing=True)
                ender = torch.cuda.Event(enable_timing=True)
                starter.record()
                _ = model(inp)
                ender.record()
                torch.cuda.synchronize()
                times.append(starter.elapsed_time(ender))  # ms
            else:
                t0 = time.perf_counter()
                _ = model(inp)
                t1 = time.perf_counter()
                times.append((t1 - t0) * 1000.0)

    return float(np.median(times))

def _default_input_hw_timm(model):
    # Try to read timm's default cfg input size (usually (3,224,224))
    try:
        cfg = getattr(model, "default_cfg", {}) or {}
        inp = cfg.get("input_size", None)
        if inp is not None and len(inp) == 3:
            return int(inp[1]), int(inp[2])
    except Exception:
        pass
    return None, None

def _make_input_that_works(model, preferred=256):
    """
    Tries preferred size first, then model default, then a few safe fallbacks.
    Returns (inp_tensor, used_size, error_if_failed).
    """
    h_def, w_def = _default_input_hw_timm(model)
    candidates = [preferred]

    if h_def and w_def and h_def == w_def:
        candidates.append(h_def)

    # safe fallbacks that often satisfy window constraints
    candidates += [224, 240, 252, 256, 280, 288, 320, 384]

    # unique, keep order
    seen = set()
    sizes = []
    for s in candidates:
        if s not in seen:
            seen.add(s)
            sizes.append(s)

    dtype = torch.float16 if (USE_FP16 and device == "cuda") else torch.float32

    last_err = None
    for s in sizes:
        try:
            inp = torch.randn(BATCH, 3, s, s, device=device, dtype=dtype)
            with torch.no_grad():
                _ = model(inp)  # just to test divisibility/shape constraints
            return inp, s, None
        except Exception as e:
            last_err = e
            continue

    return None, None, last_err

# ---------- Model Zoo (timm ids may vary by timm version) ----------
MODEL_ZOO = {
    "MobileNetV4 T": ["mobilenetv4_conv_small.e2400_r224_in1k", "mobilenetv4_tiny", "mobilenetv4_small"],
    "EfficientNetV2 B0": ["tf_efficientnetv2_b0", "efficientnetv2_rw_t"],
    "ConvNeXt V2 Tiny": ["convnextv2_tiny"],
    "MaxViT Tiny": ["maxvit_tiny_tf_224", "maxvit_tiny_tf_256", "maxvit_tiny_rw_224"],
    "CoAtNet0": ["coatnet_0_rw_224", "coatnet_0_rw_384"],
    "Swin Transformer V2 Tiny": ["swinv2_tiny_window8_256", "swinv2_tiny_window16_256", "swinv2_tiny_window8_224"],
    "DeiT III Small": ["deit3_small_patch16_224", "deit3_small_patch16_384"],
    "Vision Transformer (ViT) Small": ["vit_small_patch16_224", "vit_small_patch16_384"],
    "RegNetY 8GF": ["regnety_080"],
    "LeViT (128S)": ["levit_128s"],
    "EfficientFormerV2 S0": ["efficientformerv2_s0"],
    "MobileViT v2 XS": ["mobilevitv2_050", "mobilevitv2_075", "mobilevitv2_100"],
    "TinyViT (21M)": ["tiny_vit_21m_224"],
    "FocalNet Tiny": ["focalnet_tiny_srf", "focalnet_tiny_lrf"],
    "VAN (Visual Attention Network) B0": ["van_b0"],
    "PVTv2 B1": ["pvt_v2_b1"],
    "NextViT S": ["nextvit_small", "nextvit_s"],
    "EdgeNeXt XS": ["edgenext_x_small", "edgenext_xs"],
    "PoolFormer (MetaFormer) (S)": ["poolformer_s12", "poolformer_s24"],
    "XCiT (N12)": ["xcit_nano_12_p16_224", "xcit_nano_12_p8_224"],
    "RepVGG (A0)": ["repvgg_a0"],
    "NFNet (F0)": ["nfnet_f0"],
    "GhostNetV2": ["ghostnetv2_100", "ghostnetv2_130", "ghostnetv2_160"],
    "ConvMixer (768)": ["convmixer_768_32"],
    "MLP-Mixer (S)": ["mixer_s16_224", "mixer_s32_224"],
    "ResNet-RS (50)": ["resnetrs50"],
    "InceptionNeXt (Tiny)": ["inceptionnext_tiny"],
    "MobileOne (S0)": ["mobileone_s0"],
    "HorNet (Tiny)": ["hornet_tiny"],
}

def _try_create_timm(model_ids):
    last_err = None
    for mid in model_ids:
        try:
            m = timm.create_model(mid, pretrained=False)
            return mid, m, None
        except Exception as e:
            last_err = e
    return None, None, last_err

results = []

# ---------- Torch/timm models ----------
for disp_name, candidate_ids in MODEL_ZOO.items():
    mid, m, err = _try_create_timm(candidate_ids)
    if m is None:
        results.append({
            "Model": disp_name,
            "Params (M)": np.nan,
            "MACs (G)": np.nan,
            "FLOPs (G)": np.nan,
            "Peak GPU Mem (MB)": np.nan,
            "Latency on Tesla P100 (ms)": np.nan,
            "_status": f"SKIPPED (not found in timm): {str(err)[:120]}",
        })
        continue

    m = m.to(device).eval()

    inp, used_s, fail = _make_input_that_works(m, preferred=PREFERRED_IMG)
    if inp is None:
        results.append({
            "Model": disp_name,
            "Params (M)": np.nan,
            "MACs (G)": np.nan,
            "FLOPs (G)": np.nan,
            "Peak GPU Mem (MB)": np.nan,
            "Latency on Tesla P100 (ms)": np.nan,
            "_status": f"SKIPPED (no valid input size): {str(fail)[:120]}",
        })
        del m
        if device == "cuda":
            torch.cuda.empty_cache()
            torch.cuda.synchronize()
        continue

    # Some ops may be unstable in fp16; ensure float32 for profiling
    inp32 = inp.float()

    try:
        params_m = _params_m_torch(m)
        macs_g, flops_g = _macs_flops_torch(m, inp32)
        peak_mb = _peak_mem_torch(m, inp32)
        lat_ms = _median_latency_torch(m, inp32)

        results.append({
            "Model": disp_name,
            "Params (M)": params_m,
            "MACs (G)": macs_g,
            "FLOPs (G)": flops_g,
            "Peak GPU Mem (MB)": peak_mb,
            "Latency on Tesla P100 (ms)": lat_ms,
            "_status": f"OK (input={used_s}x{used_s}, timm_id={mid})",
        })
    except Exception as e:
        results.append({
            "Model": disp_name,
            "Params (M)": np.nan,
            "MACs (G)": np.nan,
            "FLOPs (G)": np.nan,
            "Peak GPU Mem (MB)": np.nan,
            "Latency on Tesla P100 (ms)": np.nan,
            "_status": f"FAILED during measure: {str(e)[:140]} (input={used_s}x{used_s}, timm_id={mid})",
        })

    del m
    if device == "cuda":
        torch.cuda.empty_cache()
        torch.cuda.synchronize()

# ---------- Your model (TensorFlow/Keras) ----------
import tensorflow as tf

def tf_params_m(model):
    return float(model.count_params() / 1e6)

def tf_flops(model, input_shape=(1, 256, 256, 3)):
    try:
        from tensorflow.python.framework.convert_to_constants import convert_variables_to_constants_v2

        @tf.function
        def _f(x):
            return model(x, training=False)

        concrete = _f.get_concrete_function(tf.TensorSpec(input_shape, tf.float32))
        frozen = convert_variables_to_constants_v2(concrete)
        graph = frozen.graph

        with graph.as_default():
            run_meta = tf.compat.v1.RunMetadata()
            opts = tf.compat.v1.profiler.ProfileOptionBuilder.float_operation()
            flops = tf.compat.v1.profiler.profile(graph=graph, run_meta=run_meta, cmd="op", options=opts)
        return float(flops.total_float_ops) if flops is not None else np.nan
    except Exception:
        return np.nan

def tf_latency_ms(model, input_shape=(1, 256, 256, 3), warmup=WARMUP, runs=RUNS):
    x = tf.random.normal(input_shape, dtype=tf.float32)

    @tf.function
    def _infer(t):
        return model(t, training=False)

    for _ in range(warmup):
        _ = _infer(x)
    try:
        tf.experimental.sync_devices()
    except Exception:
        pass

    times = []
    for _ in range(runs):
        t0 = time.perf_counter()
        _ = _infer(x)
        try:
            tf.experimental.sync_devices()
        except Exception:
            pass
        t1 = time.perf_counter()
        times.append((t1 - t0) * 1000.0)
    return float(np.median(times))

def tf_peak_mem_mb():
    try:
        info = tf.config.experimental.get_memory_info("GPU:0")  # bytes
        return float(info.get("peak", np.nan) / (1024**2))
    except Exception:
        return np.nan

try:
    ours_model = tf.keras.models.load_model(final_model_path, custom_objects=custom_objects, compile=False)

    ours_params_m = tf_params_m(ours_model)
    ours_flops = tf_flops(ours_model, input_shape=(1, IMG_SIZE[0], IMG_SIZE[1], 3))
    ours_flops_g = float(ours_flops / 1e9) if np.isfinite(ours_flops) else np.nan
    ours_macs_g = float(ours_flops_g / 2.0) if np.isfinite(ours_flops_g) else np.nan

    # run once then read peak (best effort)
    _ = tf_latency_ms(ours_model, input_shape=(1, IMG_SIZE[0], IMG_SIZE[1], 3), warmup=WARMUP, runs=5)
    ours_peak_mb = tf_peak_mem_mb()
    ours_lat_ms = tf_latency_ms(ours_model, input_shape=(1, IMG_SIZE[0], IMG_SIZE[1], 3), warmup=WARMUP, runs=RUNS)

    results.append({
        "Model": "PruDensNet (Ours)",
        "Params (M)": ours_params_m,
        "MACs (G)": ours_macs_g,
        "FLOPs (G)": ours_flops_g,
        "Peak GPU Mem (MB)": ours_peak_mb,
        "Latency on Tesla P100 (ms)": ours_lat_ms,
        "_status": f"OK (input={IMG_SIZE[0]}x{IMG_SIZE[1]}, tf/keras)",
    })
except Exception as e:
    results.append({
        "Model": "PruDensNet (Ours)",
        "Params (M)": np.nan,
        "MACs (G)": np.nan,
        "FLOPs (G)": np.nan,
        "Peak GPU Mem (MB)": np.nan,
        "Latency on Tesla P100 (ms)": np.nan,
        "_status": f"SKIPPED (tf measure failed): {str(e)[:140]}",
    })

# ---------- Final table ----------
df_metrics = pd.DataFrame(results)

# print ONLY requested columns (plus a status column so you know input sizes used)
show_cols = ["Model", "Params (M)", "MACs (G)", "FLOPs (G)", "Peak GPU Mem (MB)", "Latency on Tesla P100 (ms)", "_status"]
print(tabulate(df_metrics[show_cols], headers="keys", tablefmt="github", showindex=False))

# Optional: save
# df_metrics.to_csv("/kaggle/working/model_complexity_table.csv", index=False)
# print("Saved:", "/kaggle/working/model_complexity_table.csv")
```
